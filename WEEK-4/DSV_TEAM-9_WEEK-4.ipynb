```python
explicit_counts = df["explicit"].value_counts()

plt.figure(figsize=(7,5))
sns.barplot(x=explicit_counts.index.astype(str), y=explicit_counts.values, hue=explicit_counts.index.astype(str), palette="Set2", legend=False)
plt.title("Distribution of Explicit and Non-Explicit Songs", fontsize=16, fontweight="bold")
plt.xlabel("Explicit")
plt.ylabel("Number of Songs")
plt.show()
```

```python
df["duration_minutes"] = df["duration_ms.x"] / 60000
```
