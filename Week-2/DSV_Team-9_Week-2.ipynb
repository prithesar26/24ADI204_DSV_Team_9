### 7. Missing Value Analysis

Missing values occur when information is unavailable for certain records.

The `isnull()` function identifies missing values, while `sum()` calculates their total count.

Missing percentages are calculated to measure the completeness of each column.

During Week 2, missing values are only identified and documented. They will be handled during Week 3.


```python

missing_report = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().mean() * 100
    ).round(2)
})

missing_report = missing_report.sort_values(
    by="Missing Count",
    ascending=False
)

display(missing_report)

print(
    "Total Missing Cells:",
    df.isnull().sum().sum()
)

print(
    "Rows with Missing Values:",
    df.isnull().any(axis=1).sum()
)

print(
    "Columns with Missing Values:",
    (df.isnull().sum() > 0).sum()
)
```


#### Observation

The original Spotify dataset contains missing values in multiple columns.

The preview_url attribute has a relatively large number of missing values compared with several other attributes.

Some audio features and metadata columns also contain missing values.

These issues will be investigated during Week 3 to determine whether imputation, removal, or preservation is appropriate.



### 8. Duplicate Record Analysis

Duplicate records can affect the accuracy of data analysis.

The `duplicated()` function is used to identify exact duplicate rows.

Spotify song identifiers are also examined separately to identify repeated song entries.

No records are removed during Week 2.


```python

# Exact duplicate records
duplicate_rows = df.duplicated().sum()

print("Exact Duplicate Rows:", duplicate_rows)

# Identify repeated Spotify song IDs
if "id" in df.columns:
    duplicate_ids = df["id"].duplicated().sum()
    print("Repeated Spotify IDs:", duplicate_ids)
else:
    print("Column 'id' not found.")
```


#### Observation

The dataset contains no exact duplicate rows.

Repeated Spotify identifiers are examined separately because the same song may appear in multiple records.

Any repeated identifiers will be investigated before deciding whether records should be removed.



### 9. Data Type Analysis

Correct data types are important for statistical analysis and visualization.

The dataset contains numerical, categorical, and date-related attributes.

Some columns may contain numerical values stored as text.

This section identifies the current data types and checks for possible inconsistencies without modifying the original dataset.


```python

dtype_report = pd.DataFrame({
    "Column Name": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Unique Values": df.nunique(dropna=True).values
})

display(dtype_report)

print("\nData Type Distribution:")
print(df.dtypes.value_counts())
```

```python

# Check whether selected fields contain
# values that cannot be converted to numbers

numeric_candidates = [
    "popularity",
    "track_number",
    "duration_ms.x"
]

for col in numeric_candidates:
    if col in df.columns:
        converted = pd.to_numeric(
            df[col],
            errors="coerce"
        )

        invalid_count = (
            df[col].notna() & converted.isna()
        ).sum()

        print(
            f"{col}: {invalid_count} invalid numeric values"
        )
```


#### Observation

The dataset contains attributes with different data types.

Some numerical fields are stored as object data types and may require conversion.

The validation process identifies values that cannot be directly converted into numerical form.

Data type correction will be performed during Week 3.



### 10. Automated First Impression Data Quality Report

The First Impression Data Quality Report summarizes the initial condition of the Spotify dataset.

The report includes:

1. Column names
2. Data types
3. Non-null counts
4. Missing values
5. Missing percentages
6. Unique values

The report is generated automatically using Pandas and exported as a CSV file.

This report will be used to plan data cleaning activities in Week 3.


```python

quality_report = pd.DataFrame({
    "Column Name": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Non-Null Count": df.notnull().sum().values,
    "Missing Count": df.isnull().sum().values,
    "Missing Percentage": (
        df.isnull().mean() * 100
    ).round(2).values,
    "Unique Values": df.nunique(dropna=True).values
})

display(quality_report)

quality_report.to_csv(
    "Week2_Data_Quality_Report.csv",
    index=False
)

print("Data Quality Report generated successfully!")
```

```python

summary = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Missing Cells",
        "Rows with Missing Values",
        "Columns with Missing Values",
        "Exact Duplicate Rows",
        "Memory Usage (MB)"
    ],
    "Value": [
        df.shape[0],
        df.shape[1],
        int(df.isnull().sum().sum()),
        int(df.isnull().any(axis=1).sum()),
        int((df.isnull().sum() > 0).sum()),
        int(df.duplicated().sum()),
        round(
            df.memory_usage(deep=True).sum()
            / (1024 ** 2),
            2
        )
    ]
})

display(summary)

summary.to_csv(
    "Week2_Dataset_Summary.csv",
    index=False
)
```


#### Observation

The automated Data Quality Report summarizes the structure and completeness of all 46 columns.

It identifies missing values, data types, and the number of unique values in each attribute.

The overall dataset summary provides key quality metrics such as duplicate counts and missing records.

These reports provide a foundation for data cleaning and preprocessing in Week 3.
